```python
data = data[~(data["price_numeric"].isna())| ~(data["area_numeric"].isna())]
y = data["price_numeric"]
x = data["area_numeric"]
```

```python
import gurobipy as gp
```

```python
# Model

model = gp.Model("linear_regression")

# Decision variables 

b0 = model.addVar(lb=-gp.GRB.INFINITY, name="b0")
b1 = model.addVar(lb=-gp.GRB.INFINITY, name="b1")

# Objective function

errors = []

for i in range(len(x)):
    squared_error = (y[i] - (b0 + b1 * x[i]))**2
    errors.append(squared_error)

model.setObjective(gp.quicksum(errors), sense=gp.GRB.MINIMIZE)

# Optimize the model

model.optimize()    
```

```python
# Standard scaler
value_mean = data.select("chroma_cens_kurtosis_01").mean().item()
value_std = data.select("chroma_cens_kurtosis_01").std().item()


new_distribution = (data.select("chroma_cens_kurtosis_01") - value_mean )/ value_std

new_distribution.mean()
new_distribution.std()
```

```python
# Min Max scaler

value_min = data.select("chroma_cens_kurtosis_01").min().item()
value_max = data.select("chroma_cens_kurtosis_01").max().item()

new_distribution = (data.select("chroma_cens_kurtosis_01") - value_min )/ (value_max - value_min)
print(new_distribution.min())
print(new_distribution.max())
```

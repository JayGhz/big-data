```python
import numpy as np
import pandas as pd
```

```python
from sklearn.feature_extraction.text import CountVectorizer

sola_remix = ['real', 'muerte', 'real', 'muerte', 'real', 'angel', 'sola', 'sola', 'sola']
amanece = ['real', 'muerte', 'real', 'muerte', 'real', 'muerte', 'casualidad', 'sucede', 'ganas', 'ganas', 'ganas']
secreto = ['real', 'muerte', 'real', 'muerte', 'real', 'muerte', 'secreto', 'secreto', 'secreto', 'secreto']

lyrics = [' '.join(sola_remix), ' '.join(amanece), ' '.join(secreto)]

vectorizer = CountVectorizer()

frequency = vectorizer.fit_transform(lyrics).toarray()
```

```python
pd.DataFrame(frequency, columns=vectorizer.get_feature_names_out())

frequency
```

```python
N = frequency.shape[0]

# Term Frequency    
tf = frequency / np.sum(frequency, axis=1, keepdims=True)

# Document Frequency
df = np.sum(frequency > 0, axis=0)

# Inverse Document Frequency
idf = np.log(N / (df + 1))

# TF-IDF
tf_idf = tf * idf

pd.DataFrame(tf_idf, columns=vectorizer.get_feature_names_out())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf_vectorizer = TfidfVectorizer()

tfidf = tfidf_vectorizer.fit_transform(lyrics).toarray()
pd.DataFrame(tfidf, columns=tfidf_vectorizer.get_feature_names_out())
```
